```python

# ── Celda de configuración global ─────────────────────────────────────────────
# Ejecuta esta celda ANTES de cualquier diagrama.
# ──────────────────────────────────────────────────────────────────────────────
import os
import matplotlib
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import FancyBboxPatch, Ellipse
import matplotlib.patheffects as pe
import numpy as np

# ── Renderizado ───────────────────────────────────────────────────────────────
matplotlib.rcParams['font.family'] = 'DejaVu Sans'
matplotlib.rcParams['figure.dpi']  = 150       # pantalla (rápido)
matplotlib.rcParams['savefig.dpi'] = 300       # disco   (calidad imprenta)

# ── Carpeta de salida ─────────────────────────────────────────────────────────
OUTPUT_DIR = './imagenes'
os.makedirs(OUTPUT_DIR, exist_ok=True)

# ── Paleta académica SENDA — azules institucionales ───────────────────────────
# Colores primarios (usados en cabeceras, bordes y elementos de primer plano)
NAVY        = '#1A237E'   # Azul marino oscuro  → títulos, cabeceras principales
DARK_BLUE   = '#1565C0'   # Azul oscuro         → elementos primarios
ROYAL_BLUE  = '#1976D2'   # Azul real           → elementos secundarios
MED_BLUE    = '#1E88E5'   # Azul medio          → variante adicional
DEEP_BLUE   = '#0D47A1'   # Azul profundo       → énfasis / capa de datos
SLATE_BLUE  = '#283593'   # Azul pizarra        → variante oscura
INDIGO      = '#3949AB'   # Índigo              → variante media
STEEL       = '#37474F'   # Gris acero          → divisores, actores neutros

# Colores de fondo (fills claros — nunca como color principal)
LIGHT_BLUE  = '#E3F2FD'   # Azul muy claro      → fondo de elementos primarios
WHITE_BLUE  = '#E8EAF6'   # Blanco azulado      → fondo variante
PALE_BLUE   = '#BBDEFB'   # Azul pálido         → fondo secundario
SKY_BLUE    = '#B3E5FC'   # Azul cielo          → fondo módulos auxiliares

# Colores semánticos UML (se usan SOLO en el diagrama de estados)
SEM_PENDING   = '#F57F17'  # Ámbar   → estado PENDING  (esperando)
SEM_APPROVED  = '#2E7D32'  # Verde   → estado APPROVED  (aprobado)
SEM_REJECTED  = '#C62828'  # Rojo    → estado REJECTED  (rechazado)
SEM_PROCESSED = '#0D47A1'  # Azul    → estado PROCESSED (procesado)

# Tipografía y neutros
FOOTER      = '#9E9E9E'   # Gris claro  → pie de página
DARK_TEXT   = '#212121'   # Negro suave → texto principal
BODY_TEXT   = '#424242'   # Gris oscuro → texto secundario

# ── Función de guardado + visualización inline ────────────────────────────────
def save(fig, name):
    path = os.path.join(OUTPUT_DIR, name)
    fig.savefig(path, bbox_inches='tight', facecolor='white', dpi=300)
    print(f'✅  Guardado: {path}')
    plt.show()

print('✅ Configuración cargada.')
print(f'   Directorio de salida: {os.path.abspath(OUTPUT_DIR)}')
```

## Diagrama 1 — Matriz FODA

Genera una matriz de 2×2 con Fortalezas, Oportunidades, Debilidades y Amenazas.

```python

# ─────────────────────────────────────────────────────────────────────────────
# Diagrama 1 — Análisis FODA (paleta académica azul institucional)
# ─────────────────────────────────────────────────────────────────────────────

fortalezas = [
    'F1. Arquitectura moderna (React 19, Node.js, PostgreSQL)',
    'F2. RBAC con 5 niveles de acceso diferenciados',
    'F3. Actualización en tiempo real vía WebSockets',
    'F4. Nómina automática + cuentas por cobrar integradas',
    'F5. Historial de auditoría completo por registro',
    'F6. Módulo de quiosco para marcación presencial',
]
debilidades = [
    'D1. Requiere conectividad a internet constante',
    'D2. Curva de aprendizaje para usuarios no digitalizados',
    'D3. Sin aplicación móvil nativa para quiosco',
    'D4. Configuración inicial de cuentas contables',
    '       requiere personal especializado',
]
oportunidades = [
    'O1. Extensión a otras instituciones adventistas',
    'O2. Integración con nómina institucional existente',
    'O3. Adopción de app móvil para el quiosco',
    'O4. Analítica avanzada en reportes gerenciales',
    'O5. Exportación a estándares contables nacionales',
]
amenazas = [
    'A1. Cambios regulatorios en políticas de UNADECA',
    'A2. Resistencia al cambio del personal administrativo',
    'A3. Interrupciones del servicio en la nube (Supabase)',
    'A4. Vulnerabilidades de seguridad en dependencias',
    'A5. Pérdida de datos por falla de conectividad',
]

# ── Paleta por cuadrante: solo azules institucionales + gris acero ─────────────
# (hc=header, bg=background, accent=barra lateral)
quads = [
    (NAVY,       WHITE_BLUE, PALE_BLUE,  fortalezas,    'FORTALEZAS',    'Origen interno · Aspecto positivo'),
    (STEEL,      '#ECEFF1',  '#B0BEC5',  debilidades,   'DEBILIDADES',   'Origen interno · Aspecto negativo'),
    (DARK_BLUE,  LIGHT_BLUE, '#90CAF9',  oportunidades, 'OPORTUNIDADES', 'Origen externo · Aspecto positivo'),
    (SLATE_BLUE, WHITE_BLUE, '#9FA8DA',  amenazas,      'AMENAZAS',      'Origen externo · Aspecto negativo'),
]

fig = plt.figure(figsize=(16, 11))
fig.patch.set_facecolor('#F5F5F5')

# ── Franja de título superior ─────────────────────────────────────────────────
title_ax = fig.add_axes([0.0, 0.92, 1.0, 0.08])
title_ax.set_xlim(0, 1)
title_ax.set_ylim(0, 1)
title_ax.axis('off')
title_ax.add_patch(FancyBboxPatch((0, 0), 1, 1,
    boxstyle='square,pad=0', fc=NAVY, ec='none'))
title_ax.text(0.5, 0.64, 'ANÁLISIS FODA — SISTEMA SENDA',
    ha='center', va='center', fontsize=17, fontweight='bold', color='white')
title_ax.text(0.5, 0.20, 'Universidad Adventista de Centro América (UNADECA) · 2026',
    ha='center', va='center', fontsize=10, color='#B0BEC5')

# ── Etiquetas de eje: POSITIVO / NEGATIVO y INTERNO / EXTERNO ─────────────────
label_ax = fig.add_axes([0.0, 0.04, 1.0, 0.88])
label_ax.set_xlim(0, 1)
label_ax.set_ylim(0, 1)
label_ax.axis('off')

for xc, txt in [(0.28, 'POSITIVO'), (0.74, 'NEGATIVO')]:
    label_ax.text(xc, 0.975, txt,
        ha='center', va='top', fontsize=12, fontweight='bold',
        color=STEEL, alpha=0.35)

for yc, txt in [(0.78, 'INTERNO'), (0.30, 'EXTERNO')]:
    label_ax.text(0.008, yc, txt,
        ha='left', va='center', fontsize=10, fontweight='bold',
        color=STEEL, alpha=0.35, rotation=90)

# Línea divisoria central (horizontal y vertical)
label_ax.plot([0.50, 0.50], [0.02, 0.98], color=STEEL, lw=2, alpha=0.25)
label_ax.plot([0.015, 0.99], [0.525, 0.525], color=STEEL, lw=2, alpha=0.25)

# ── 4 cuadrantes ──────────────────────────────────────────────────────────────
# Orden: Fortalezas (sup-izq), Debilidades (sup-der),
#        Oportunidades (inf-izq), Amenazas (inf-der)
positions = [
    (0.020, 0.535, 0.475, 0.44),   # Fortalezas
    (0.515, 0.535, 0.475, 0.44),   # Debilidades
    (0.020, 0.055, 0.475, 0.44),   # Oportunidades
    (0.515, 0.055, 0.475, 0.44),   # Amenazas
]

for (lx, ly, lw_ax, lh_ax), (hc, bg, accent, items, title, subtitle) in zip(positions, quads):
    q_ax = fig.add_axes([lx, ly, lw_ax, lh_ax])
    q_ax.set_xlim(0, 1)
    q_ax.set_ylim(0, 1)
    q_ax.axis('off')
    q_ax.set_facecolor(bg)

    # Borde exterior redondeado
    q_ax.add_patch(FancyBboxPatch((0.005, 0.005), 0.99, 0.99,
        boxstyle='round,pad=0.01', fc='none', ec=hc, lw=2.0))

    # Franja de cabecera
    q_ax.add_patch(mpatches.Rectangle((0, 0.82), 1, 0.18, fc=hc, ec='none'))

    # Barra lateral de acento
    q_ax.add_patch(mpatches.Rectangle((0, 0), 0.025, 0.82,
        fc=accent, ec='none', alpha=0.6))

    # Título y subtítulo del cuadrante
    q_ax.text(0.5, 0.915, title,
        ha='center', va='center', fontsize=10, fontweight='bold', color='white')
    q_ax.text(0.5, 0.847, subtitle,
        ha='center', va='center', fontsize=7.5, color='white', alpha=0.85)

    # Ítems con bullet circular
    y_pos = 0.775
    for item in items:
        q_ax.add_patch(mpatches.Circle((0.048, y_pos + 0.007), 0.012,
            fc=hc, ec='none', alpha=0.8))
        q_ax.text(0.070, y_pos, item,
            ha='left', va='center', fontsize=8.2, color=DARK_TEXT)
        y_pos -= 0.133

# ── Pie de página ─────────────────────────────────────────────────────────────
fig.text(0.5, 0.010,
    'Sistema SENDA — UNADECA 2026  |  Documento de Trabajo de Graduación',
    ha='center', fontsize=7.5, color=FOOTER, style='italic')

save(fig, '1_foda.png')
```

## Diagrama 2 — Diagrama de Arquitectura del Sistema (3 capas)

**Dónde va en el Word:** Sección **11. Productos y Resultados Esperados**

Muestra las tres capas: Presentación (SPA), Lógica de Negocio (API REST) y Datos (Supabase).

```python

# ─────────────────────────────────────────────────────────────────────────────
# Diagrama 2 — Arquitectura del Sistema en 3 Capas (paleta azul académica)
# ─────────────────────────────────────────────────────────────────────────────

fig, ax = plt.subplots(figsize=(18, 12))
ax.set_xlim(0, 18)
ax.set_ylim(0, 12)
ax.axis('off')
ax.set_facecolor('#F5F5F5')
fig.patch.set_facecolor('#F5F5F5')

# ── Helpers ───────────────────────────────────────────────────────────────────
def layer_band(ax, y_bot, height, fc, ec, label, sublabel):
    """Franja de capa con cabecera coloreada y subtítulo tecnológico."""
    ax.add_patch(FancyBboxPatch((0.35, y_bot), 17.3, height,
        boxstyle='round,pad=0.12', fc=fc, ec=ec, lw=2.0, zorder=1))
    ax.add_patch(mpatches.Rectangle((0.35, y_bot + height - 0.58), 17.3, 0.58,
        fc=ec, ec='none', zorder=2))
    ax.text(9.0, y_bot + height - 0.28, label,
        ha='center', va='center', fontsize=11, fontweight='bold', color='white', zorder=3)
    ax.text(9.0, y_bot + height - 0.52, sublabel,
        ha='center', va='center', fontsize=7.5, color='white', alpha=0.85, zorder=3)

def comp_box(ax, x, y, w, h, title, tech, fc, ec):
    """Caja de componente con título + badge de tecnología."""
    ax.add_patch(FancyBboxPatch((x, y), w, h,
        boxstyle='round,pad=0.06', fc=fc, ec=ec, lw=1.8, zorder=4,
        path_effects=[pe.withSimplePatchShadow(
            offset=(2, -2), shadow_rgbFace='#90A4AE', alpha=0.4)]))
    ax.text(x + w/2, y + h * 0.62, title,
        ha='center', va='center', fontsize=8.0, fontweight='bold', color=DARK_TEXT, zorder=5)
    ax.add_patch(FancyBboxPatch((x + w*0.1, y + 0.06), w*0.8, 0.28,
        boxstyle='round,pad=0.03', fc=ec, ec='none', alpha=0.22, zorder=5))
    ax.text(x + w/2, y + 0.20, tech,
        ha='center', va='center', fontsize=6.5, color=ec, fontweight='bold', zorder=6)

def arrow_v(ax, x, y_top, y_bot, label_up, label_dn, color):
    """Flecha bidireccional vertical entre capas."""
    ax.annotate('', xy=(x, y_bot + 0.05), xytext=(x, y_top - 0.05),
        arrowprops=dict(arrowstyle='->', color=color, lw=2.0))
    ax.annotate('', xy=(x, y_top - 0.05), xytext=(x, y_bot + 0.05),
        arrowprops=dict(arrowstyle='->', color=color, lw=2.0))
    ax.text(x + 0.20, (y_top + y_bot) / 2 + 0.10, label_up,
        ha='left', va='center', fontsize=7.5, color=color, fontweight='bold')
    ax.text(x + 0.20, (y_top + y_bot) / 2 - 0.22, label_dn,
        ha='left', va='center', fontsize=7.0, color=color, style='italic')

# ── Título superior ───────────────────────────────────────────────────────────
ax.add_patch(mpatches.Rectangle((0, 11.35), 18, 0.65, fc=NAVY, ec='none', zorder=0))
ax.text(9.0, 11.67, 'ARQUITECTURA DEL SISTEMA — SENDA',
    ha='center', va='center', fontsize=14, fontweight='bold', color='white', zorder=1)
ax.text(9.0, 11.40, 'Patrón de 3 capas: Presentación · Lógica de Negocio · Datos',
    ha='center', va='center', fontsize=8.5, color='#B0BEC5', zorder=1)

# ── Capa 1 — Presentación (SPA) ───────────────────────────────────────────────
layer_band(ax, 8.45, 2.65, WHITE_BLUE, NAVY,
    'CAPA 1  —  PRESENTACIÓN  (SPA)',
    'React 19 · TypeScript · Vite · Tailwind CSS  |  Desplegada en Docker / nginx')

portals = [
    ('Portal\nEstudiante',    'React', '#C5CAE9', INDIGO,     0.55),
    ('Portal\nJefe Dpto.',     'React', '#C5CAE9', INDIGO,     3.25),
    ('Portal\nAdministrador', 'React', '#C5CAE9', INDIGO,     5.95),
    ('Portal\nSuperAdmin',    'React', '#C5CAE9', INDIGO,     8.65),
    ('Portal\nContabilidad',  'React', '#C5CAE9', INDIGO,    11.35),
    ('Módulo\nQuiosco',       'React', PALE_BLUE, DARK_BLUE, 14.05),
]
for title, tech, fc, ec, px in portals:
    comp_box(ax, px, 8.80, 2.5, 1.90, title, tech, fc, ec)

# ── Capa 2 — Lógica de Negocio (API REST) ─────────────────────────────────────
layer_band(ax, 4.30, 3.80, LIGHT_BLUE, DARK_BLUE,
    'CAPA 2  —  LÓGICA DE NEGOCIO  (API REST)',
    'Node.js 20 LTS · Express 4 · ESM  |  Desplegada en Docker · senda-backend')

features = [
    ('auth',        'JWT\nrequireAuth',  '#90CAF9', DARK_BLUE),
    ('users',       'RBAC\n5 roles',     '#90CAF9', DARK_BLUE),
    ('departments', 'CRUD\nDeptos.',     '#90CAF9', DARK_BLUE),
    ('rates',       'Tarifa\nHora',      '#90CAF9', DARK_BLUE),
    ('work_logs',   'Registro\nHoras',   PALE_BLUE, DEEP_BLUE),
    ('kiosk',       'Sesión\nQuiosco',   PALE_BLUE, DEEP_BLUE),
    ('reports',     'PDF / CSV\nNómina', SKY_BLUE,  SLATE_BLUE),
    ('accounting',  'Config.\nContable', SKY_BLUE,  SLATE_BLUE),
]
fw = 1.90; gap = 0.25; fx0 = 0.55
for i, (name, tech, fc, ec) in enumerate(features):
    comp_box(ax, fx0 + i * (fw + gap), 4.65, fw, 1.85, name, tech, fc, ec)

# Banner de middleware
ax.add_patch(FancyBboxPatch((0.55, 6.70), 17.1, 0.38,
    boxstyle='round,pad=0.04', fc=WHITE_BLUE, ec=DARK_BLUE, lw=1.0, alpha=0.7, zorder=3))
ax.text(9.0, 6.89,
    'Middleware: requireAuth  →  getRequesterProfile  →  Feature Router  →  Error Handler',
    ha='center', va='center', fontsize=8, color=SLATE_BLUE, style='italic', zorder=4)

# ── Capa 3 — Datos (Supabase) ─────────────────────────────────────────────────
layer_band(ax, 0.50, 3.50, PALE_BLUE, DEEP_BLUE,
    'CAPA 3  —  DATOS  (Supabase Self-Hosted)',
    'PostgreSQL 15 · Auth · Realtime · Row Level Security (RLS)')

db_comps = [
    ('PostgreSQL\nTablas principales',
     'profiles · departments\nwork_logs · rates',
     '#BBDEFB', DEEP_BLUE, 0.55, 4.1),
    ('PostgreSQL\nTablas quiosco / contabilidad',
     'kiosk_state · kiosk_sessions\naccounting_config · receivables',
     '#BBDEFB', DEEP_BLUE, 5.35, 4.1),
    ('Supabase\nAuth',
     'JWT · signIn\ngetUser · RLS ctx',
     SKY_BLUE, SLATE_BLUE, 10.15, 3.3),
    ('Supabase\nRealtime',
     'WebSockets\nLISTEN / NOTIFY',
     SKY_BLUE, SLATE_BLUE, 13.30, 3.3),
    ('Row Level\nSecurity',
     'Políticas por rol\npor tabla',
     LIGHT_BLUE, DARK_BLUE, 16.00, 2.8),
]
for title, tech, fc, ec, px, pw in db_comps:
    comp_box(ax, px, 0.85, pw, 2.35, title, tech, fc, ec)

# ── Flechas entre capas ───────────────────────────────────────────────────────
arrow_v(ax, 9.0, 8.45, 8.10, '  HTTPS + JWT Bearer', '  Token en localStorage', INDIGO)
arrow_v(ax, 9.0, 4.30, 3.95, '  supabase-js (service role)', '  SQL / RLS bypass', DEEP_BLUE)

# ── Pie de página ─────────────────────────────────────────────────────────────
fig.text(0.5, 0.005,
    'Sistema SENDA — UNADECA 2026  |  Documento de Trabajo de Graduación',
    ha='center', fontsize=7.5, color=FOOTER, style='italic')

ax.set_title('Figura 2. Diagrama de Arquitectura del Sistema — SENDA (UNADECA, 2026)',
    fontsize=13, fontweight='bold', pad=8, color=DARK_TEXT)

save(fig, '2_arquitectura.png')
```

```python

# ─────────────────────────────────────────────────────────────────────────────
# Diagrama 4 — Diagrama de Estados: Ciclo de Vida del WorkLog (UML)
# Nota: se mantienen los colores semánticos UML estándar para los estados
#       (ámbar=espera, verde=aprobado, rojo=rechazado, azul=procesado).
# ─────────────────────────────────────────────────────────────────────────────

fig, ax = plt.subplots(figsize=(18, 9))
ax.set_xlim(0, 18)
ax.set_ylim(0, 9)
ax.axis('off')
ax.set_facecolor('#FAFAFA')
fig.patch.set_facecolor('#FAFAFA')

# ── Título ────────────────────────────────────────────────────────────────────
ax.add_patch(mpatches.Rectangle((0, 8.35), 18, 0.65, fc=NAVY, ec='none', zorder=0))
ax.text(9.0, 8.67,
    'DIAGRAMA DE ESTADOS — CICLO DE VIDA DEL REGISTRO DE HORAS (WorkLog)',
    ha='center', va='center', fontsize=13, fontweight='bold', color='white', zorder=1)
ax.text(9.0, 8.40, 'Sistema SENDA · UNADECA 2026',
    ha='center', va='center', fontsize=8, color='#B0BEC5', zorder=1)

# ── Helpers ───────────────────────────────────────────────────────────────────
def state_box(ax, cx, cy, name, stereotype, desc, fc, ec):
    """Caja de estado UML: cabecera coloreada + estereotipo + descripción."""
    w, h = 3.6, 2.0
    ax.add_patch(FancyBboxPatch((cx - w/2, cy - h/2), w, h,
        boxstyle='round,pad=0.15', fc=fc, ec=ec, lw=2.5, zorder=3,
        path_effects=[pe.withSimplePatchShadow(
            offset=(3, -3), shadow_rgbFace='#90A4AE', alpha=0.35)]))
    # Cabecera de color
    ax.add_patch(mpatches.Rectangle((cx - w/2, cy + h/2 - 0.55), w, 0.55,
        fc=ec, ec='none', zorder=4))
    ax.text(cx, cy + h/2 - 0.27, name,
        ha='center', va='center', fontsize=11.5, fontweight='bold', color='white', zorder=5)
    # Línea divisoria
    ax.plot([cx - w/2, cx + w/2], [cy + h/2 - 0.55, cy + h/2 - 0.55],
        color=ec, lw=1.0, zorder=4)
    # Estereotipo (cursiva)
    ax.text(cx, cy + 0.10, stereotype,
        ha='center', va='center', fontsize=7.5, color=ec, style='italic', zorder=5)
    # Descripción indentada
    ax.text(cx, cy - 0.38, desc,
        ha='center', va='center', fontsize=8, color=BODY_TEXT,
        multialignment='center', zorder=5)
    return {
        'left': cx - w/2, 'right': cx + w/2,
        'top':  cy + h/2, 'bottom': cy - h/2,
        'cx': cx, 'cy': cy,
    }

def transition(ax, x1, y1, x2, y2, label, guard, color, rad=0.0):
    """Flecha de transición con etiqueta de evento y condición de guarda."""
    ax.annotate('', xy=(x2, y2), xytext=(x1, y1),
        arrowprops=dict(arrowstyle='->', color=color, lw=2.0,
            connectionstyle=f'arc3,rad={rad}'))
    mx, my = (x1 + x2) / 2, (y1 + y2) / 2
    ax.text(mx, my + 0.22, label,
        ha='center', va='bottom', fontsize=8.5, fontweight='bold', color=color)
    if guard:
        ax.text(mx, my - 0.02, f'[{guard}]',
            ha='center', va='bottom', fontsize=7.5, color=color, style='italic')

# ── Pseudoestado inicial ──────────────────────────────────────────────────────
ax.add_patch(mpatches.Circle((1.5, 4.5), 0.28, fc=DARK_TEXT, ec=DARK_TEXT, zorder=5))
ax.annotate('', xy=(2.7, 4.5), xytext=(1.80, 4.5),
    arrowprops=dict(arrowstyle='->', color=DARK_TEXT, lw=2.0))
ax.text(2.10, 4.75, 'crear()',
    fontsize=7.5, color=STEEL, style='italic', ha='center', va='bottom')

# ── Estados (colores semánticos UML estándar) ─────────────────────────────────
S = {}
S['PENDING']   = state_box(ax,  4.5, 4.5, 'PENDING',   '«estado inicial»',
    'Registro creado.\nEsperando revisión\ndel Jefe de Depto.',
    '#FFF9C4', SEM_PENDING)

S['APPROVED']  = state_box(ax,  9.5, 6.5, 'APPROVED',  '«aprobado»',
    'Horas validadas.\nListo para ser\nprocesado en nómina.',
    '#C8E6C9', SEM_APPROVED)

S['REJECTED']  = state_box(ax,  9.5, 2.5, 'REJECTED',  '«rechazado»',
    'Registro no válido.\nEl estudiante puede\nreenviarlo corregido.',
    '#FFCDD2', SEM_REJECTED)

S['PROCESSED'] = state_box(ax, 14.7, 6.5, 'PROCESSED', '«procesado»',
    'Incluido en nómina.\nRegistro cerrado,\nno modificable.',
    PALE_BLUE, SEM_PROCESSED)

# ── Transiciones ──────────────────────────────────────────────────────────────
transition(ax, 6.30, 5.20, 7.70, 6.10,
    'aprobar()',   'role == DEPT_HEAD',   SEM_APPROVED,  rad=-0.1)
transition(ax, 6.30, 3.80, 7.70, 2.90,
    'rechazar()',  'role == DEPT_HEAD',   SEM_REJECTED,  rad=0.1)
transition(ax, 11.30, 6.50, 12.90, 6.50,
    'procesar()',  'role == ACCOUNTING',  SEM_PROCESSED, rad=0.0)

# Vuelta de REJECTED → PENDING (re-envío)
ax.annotate('', xy=(4.5, 3.50), xytext=(7.70, 2.10),
    arrowprops=dict(arrowstyle='->', color=STEEL, lw=1.8,
        connectionstyle='arc3,rad=0.35'))
ax.text(5.5, 2.00, 'reenviar()',
    ha='center', va='top', fontsize=8.5, fontweight='bold', color=STEEL)
ax.text(5.5, 1.72, '[estudiante corrige]',
    ha='center', va='top', fontsize=7.5, color=STEEL, style='italic')

# ── Pseudoestado final ────────────────────────────────────────────────────────
ax.add_patch(mpatches.Circle((16.75, 6.50), 0.30, fc=DARK_TEXT, ec=DARK_TEXT, zorder=5))
ax.add_patch(mpatches.Circle((16.75, 6.50), 0.44, fc='none', ec=DARK_TEXT, lw=2.2, zorder=4))
ax.annotate('', xy=(16.44, 6.50), xytext=(16.50, 6.50),
    arrowprops=dict(arrowstyle='->', color=DARK_TEXT, lw=1.5))

# ── Leyenda ───────────────────────────────────────────────────────────────────
legend_x, legend_y = 0.4, 1.5
ax.add_patch(FancyBboxPatch((legend_x, legend_y), 3.2, 1.7,
    boxstyle='round,pad=0.08', fc=WHITE_BLUE, ec=NAVY, lw=1.2, alpha=0.85, zorder=3))
ax.text(legend_x + 1.60, legend_y + 1.50, 'Convenciones UML',
    ha='center', va='center', fontsize=8, fontweight='bold', color=NAVY)
legend_items = [
    ('● Pseudoestado inicial', DARK_TEXT),
    ('⊙ Pseudoestado final',   DARK_TEXT),
    ('[guard]  condición',     STEEL),
    ('evento()  disparador',   BODY_TEXT),
]
for j, (txt, col) in enumerate(legend_items):
    ax.text(legend_x + 0.15, legend_y + 1.22 - j * 0.32, txt,
        ha='left', va='center', fontsize=7.5, color=col)

# ── Pie de página ─────────────────────────────────────────────────────────────
fig.text(0.5, 0.005,
    'Sistema SENDA — UNADECA 2026  |  Documento de Trabajo de Graduación',
    ha='center', fontsize=7.5, color=FOOTER, style='italic')

ax.set_title(
    'Figura 4. Diagrama de Estados — Ciclo de Vida del WorkLog (UNADECA, 2026)',
    fontsize=13, fontweight='bold', pad=8)

save(fig, '4_estados_worklog.png')
```

## Diagrama 5 — Diagrama Entidad-Relación (ER)

```python

# ─────────────────────────────────────────────────────────────────────────────
# Diagrama 5 — Diagrama Entidad-Relación (8 entidades, paleta azul académica)
# ─────────────────────────────────────────────────────────────────────────────

ROW_H = 0.34   # Altura de cada fila de campo

def er_entity(ax, x, y, title, pk, fields, hc, bg, w=4.6):
    """Entidad ER: cabecera + PK + campos con tipo de dato y marcador FK."""
    n_rows  = 1 + len(fields)
    total_h = ROW_H * (n_rows + 1.2)
    bottom  = y - total_h

    # Sombra decorativa
    ax.add_patch(FancyBboxPatch((x + 0.10, bottom - 0.10), w, total_h,
        boxstyle='round,pad=0.04', fc='#90A4AE', ec='none', alpha=0.25, zorder=1))
    # Cuerpo
    ax.add_patch(FancyBboxPatch((x, bottom), w, total_h,
        boxstyle='round,pad=0.04', fc=bg, ec=hc, lw=2.2, zorder=2))
    # Cabecera
    hdr_h = ROW_H * 1.2
    ax.add_patch(mpatches.Rectangle((x, y - hdr_h), w, hdr_h,
        fc=hc, ec='none', zorder=3))
    ax.text(x + w/2, y - hdr_h/2, title,
        ha='center', va='center', fontsize=9, fontweight='bold', color='white', zorder=4)

    # Clave primaria
    pk_y = y - hdr_h - ROW_H
    ax.plot([x + 0.08, x + w - 0.08], [pk_y + ROW_H, pk_y + ROW_H],
        color=hc, lw=0.5, alpha=0.5, zorder=3)
    ax.text(x + 0.15, pk_y + ROW_H * 0.5, f'PK  {pk}',
        ha='left', va='center', fontsize=7, fontweight='bold', color='#4E342E', zorder=4)

    # Campos (con prefijo FK si aplica)
    for i, (fname, ftype, is_fk) in enumerate(fields):
        ry = pk_y - (i + 1) * ROW_H
        ax.plot([x + 0.08, x + w - 0.08], [ry + ROW_H, ry + ROW_H],
            color=hc, lw=0.4, alpha=0.30, zorder=3)
        prefix = 'FK ' if is_fk else '   '
        ax.text(x + 0.12, ry + ROW_H * 0.5, f'{prefix}{fname}',
            ha='left', va='center', fontsize=6.8, color=DARK_TEXT, zorder=4)
        ax.text(x + w - 0.10, ry + ROW_H * 0.5, ftype,
            ha='right', va='center', fontsize=6.3, color='#757575',
            style='italic', zorder=4)

    cx = x + w / 2
    return {
        'cx': cx, 'top': y, 'bottom': bottom,
        'left': x, 'right': x + w, 'mid_y': (y + bottom) / 2,
    }

fig, ax = plt.subplots(figsize=(24, 16))
ax.set_xlim(0, 24)
ax.set_ylim(0, 16)
ax.axis('off')
ax.set_facecolor('#F5F5F5')
fig.patch.set_facecolor('#F5F5F5')

# ── Título ────────────────────────────────────────────────────────────────────
ax.add_patch(mpatches.Rectangle((0, 15.35), 24, 0.65, fc=NAVY, ec='none', zorder=0))
ax.text(12.0, 15.67, 'DIAGRAMA ENTIDAD-RELACIÓN — BASE DE DATOS SENDA',
    ha='center', va='center', fontsize=14, fontweight='bold', color='white', zorder=1)
ax.text(12.0, 15.40, 'PostgreSQL 15 · Supabase Self-Hosted · UNADECA 2026',
    ha='center', va='center', fontsize=8.5, color='#B0BEC5', zorder=1)

# ── Zonas de dominio (todas en azules institucionales) ────────────────────────
zones = [
    (0.2,  7.5, 11.2, 7.5, WHITE_BLUE, NAVY,       'Identidad & Organización'),
    (11.6, 7.5, 12.0, 7.5, LIGHT_BLUE, DARK_BLUE,  'Registro Laboral & Tarifas'),
    (0.2,  0.3, 11.4, 7.0, PALE_BLUE,  DEEP_BLUE,  'Módulo Quiosco'),
    (11.8, 0.3, 11.8, 7.0, SKY_BLUE,   SLATE_BLUE, 'Módulo Contabilidad'),
]
for zx, zy, zw, zh, zfc, zec, zlbl in zones:
    ax.add_patch(FancyBboxPatch((zx, zy), zw, zh,
        boxstyle='round,pad=0.1', fc=zfc, ec=zec, lw=1.2, alpha=0.30, zorder=0))
    ax.text(zx + zw/2, zy + zh - 0.18, zlbl,
        ha='center', va='center', fontsize=8, color=zec,
        fontweight='bold', alpha=0.60, zorder=1)

# ── Entidades Fila 1 — Identidad, Trabajo, Tarifas ────────────────────────────
E = {}
E['profiles'] = er_entity(ax, 0.35, 14.8, 'profiles', 'id : UUID',
    [('email',                'TEXT',    False),
     ('full_name',            'TEXT',    False),
     ('role',                 'UserRole',False),
     ('department_id',        'UUID',    True),
     ('is_active',            'BOOL',    False),
     ('must_change_password', 'BOOL',    False),
     ('hourly_rate_override', 'NUMERIC', False)],
    NAVY, WHITE_BLUE, w=5.0)

E['departments'] = er_entity(ax, 5.7, 14.8, 'departments', 'id : UUID',
    [('name',             'TEXT', False),
     ('cost_center',      'TEXT', False),
     ('kiosk_active',     'BOOL', False),
     ('kiosk_shift_start','TIME', False),
     ('kiosk_shift_end',  'TIME', False)],
    DARK_BLUE, LIGHT_BLUE, w=4.8)

E['work_logs'] = er_entity(ax, 11.0, 14.8, 'work_logs', 'id : UUID',
    [('student_id',    'UUID',        True),
     ('department_id', 'UUID',        True),
     ('start_time',    'TIMESTAMPTZ', False),
     ('end_time',      'TIMESTAMPTZ', False),
     ('hours_worked',  'NUMERIC',     False),
     ('status',        'WorkLogStatus',False),
     ('description',   'TEXT',        False),
     ('approved_by',   'UUID',        True)],
    ROYAL_BLUE, LIGHT_BLUE, w=5.2)

E['rates'] = er_entity(ax, 16.7, 14.8, 'rates', 'id : UUID',
    [('hourly_rate',    'NUMERIC', False),
     ('effective_date', 'DATE',    False),
     ('created_by',     'UUID',    True)],
    DEEP_BLUE, PALE_BLUE, w=4.6)

# ── Entidades Fila 2 — Quiosco y Contabilidad ─────────────────────────────────
E['kiosk_state'] = er_entity(ax, 0.35, 6.8, 'kiosk_state', 'id : UUID',
    [('department_id', 'UUID', True),
     ('is_active',     'BOOL', False),
     ('activated_by',  'UUID', True),
     ('session_label', 'TEXT', False)],
    SLATE_BLUE, WHITE_BLUE, w=4.8)

E['kiosk_sessions'] = er_entity(ax, 5.5, 6.8, 'kiosk_sessions', 'id : UUID',
    [('student_id',    'UUID',        True),
     ('department_id', 'UUID',        True),
     ('clock_in',      'TIMESTAMPTZ', False),
     ('clock_out',     'TIMESTAMPTZ', False),
     ('work_log_id',   'UUID',        True)],
    INDIGO, WHITE_BLUE, w=5.0)

E['accounting_config'] = er_entity(ax, 11.0, 6.8, 'accounting_config', 'id : UUID',
    [('department_id',  'UUID', True),
     ('cost_center',    'TEXT', False),
     ('account_debit',  'TEXT', False),
     ('account_credit', 'TEXT', False),
     ('tithe_account',  'TEXT', False),
     ('closing_day',    'INT',  False)],
    '#0277BD', SKY_BLUE, w=5.2)

E['receivables'] = er_entity(ax, 16.7, 6.8, 'receivables', 'id : UUID',
    [('student_id',    'UUID',    True),
     ('amount',        'NUMERIC', False),
     ('reason',        'TEXT',    False),
     ('billing_cycle', 'TEXT',    False),
     ('is_paid',       'BOOL',    False)],
    '#01579B', SKY_BLUE, w=4.6)

# ── Relaciones ────────────────────────────────────────────────────────────────
def er_rel(ax, e1, e2, verb, card1, card2, c, rad=0.0, from_s='bottom', to_s='top'):
    """Dibuja una relación entre dos entidades ER con cardinalidad y etiqueta."""
    b1, b2 = E[e1], E[e2]
    pts  = {
        'bottom': (b1['cx'],    b1['bottom']),
        'top':    (b1['cx'],    b1['top']),
        'right':  (b1['right'], b1['mid_y']),
        'left':   (b1['left'],  b1['mid_y']),
    }
    pts2 = {
        'bottom': (b2['cx'],    b2['bottom']),
        'top':    (b2['cx'],    b2['top']),
        'right':  (b2['right'], b2['mid_y']),
        'left':   (b2['left'],  b2['mid_y']),
    }
    sx, sy = pts[from_s]
    ex, ey = pts2[to_s]
    ax.annotate('', xy=(ex, ey), xytext=(sx, sy),
        arrowprops=dict(arrowstyle='->', color=c, lw=1.8,
            connectionstyle=f'arc3,rad={rad}'))
    mx, my = (sx + ex) / 2, (sy + ey) / 2
    ax.text(mx + 0.15, my + 0.12, verb,
        fontsize=6.8, color=c, style='italic', ha='left', va='bottom', zorder=5)
    ax.text(sx + 0.12, sy + 0.10, card1,
        fontsize=7.5, fontweight='bold', color=c, ha='left', va='bottom', zorder=5)
    ax.text(ex + 0.12, ey + 0.10, card2,
        fontsize=7.5, fontweight='bold', color=c, ha='left', va='bottom', zorder=5)

er_rel(ax, 'profiles',         'departments',    'pertenece a',    '0..N', '1',
    NAVY,       from_s='right', to_s='left')
er_rel(ax, 'work_logs',        'profiles',       'realizado por',  'N',    '1',
    ROYAL_BLUE, from_s='left',  to_s='right',  rad=0.25)
er_rel(ax, 'work_logs',        'departments',    'en departamento','N',    '1',
    DARK_BLUE,  from_s='left',  to_s='right',  rad=-0.15)
er_rel(ax, 'kiosk_state',      'departments',    'pertenece a',    '1',    '1',
    SLATE_BLUE, from_s='right', to_s='left',   rad=0.2)
er_rel(ax, 'kiosk_sessions',   'kiosk_state',    'sesion activa',  'N',    '1',
    INDIGO,     from_s='left',  to_s='right',  rad=0.15)
er_rel(ax, 'kiosk_sessions',   'work_logs',      'genera',         '1', '0..1',
    STEEL,      from_s='top',   to_s='bottom', rad=0.2)
er_rel(ax, 'accounting_config','departments',    'configura',      '1',    '1',
    '#0277BD',  from_s='right', to_s='left',   rad=0.0)
er_rel(ax, 'receivables',      'profiles',       'pertenece a',    'N',    '1',
    '#01579B',  from_s='right', to_s='bottom', rad=-0.12)

# ── Leyenda ───────────────────────────────────────────────────────────────────
leg_x, leg_y = 21.5, 3.5
ax.add_patch(FancyBboxPatch((leg_x, leg_y), 2.3, 2.2,
    boxstyle='round,pad=0.10', fc='white', ec=NAVY, lw=1.2, zorder=4))
ax.text(leg_x + 1.15, leg_y + 2.00, 'Leyenda',
    ha='center', fontsize=8, fontweight='bold', color=NAVY)
legend_items = [
    ('PK',    'Clave Primaria'),
    ('FK',    'Clave Foránea'),
    ('1 : N', 'Cardinalidad'),
]
for j, (icon, desc) in enumerate(legend_items):
    ax.text(leg_x + 0.15, leg_y + 1.65 - j * 0.50,
        f'{icon}  {desc}',
        ha='left', va='center', fontsize=7.2, color=BODY_TEXT, fontweight='bold')

# ── Pie de página ─────────────────────────────────────────────────────────────
fig.text(0.5, 0.005,
    'Sistema SENDA — UNADECA 2026  |  Documento de Trabajo de Graduación',
    ha='center', fontsize=7.5, color=FOOTER, style='italic')

ax.set_title('Figura 5. Diagrama Entidad-Relación — Base de Datos SENDA (UNADECA, 2026)',
    fontsize=13, fontweight='bold', pad=8)

save(fig, '5_er_diagrama.png')
```

## Diagrama 7 — Diagrama de Clases del Dominio

**Dónde va en el Word:** Sección **11. Productos y Resultados Esperados**

Dibuja con matplotlib puro las clases del dominio SENDA con atributos, métodos y relaciones UML.


```python

# ─────────────────────────────────────────────────────────────────────────────
# Diagrama 7 — Diagrama de Clases (8 clases, paleta azul académica)
# ─────────────────────────────────────────────────────────────────────────────

CLS_W = 4.4   # Ancho estándar de caja de clase
ROW_H = 0.32  # Altura de fila de atributo / método

def class_box(ax, x, y, name, stereotype, attrs, methods, hc, bg, w=CLS_W):
    """
    Caja de clase UML: cabecera coloreada + sección de atributos + sección de métodos.
    Visibilidad: '+' público, '-' privado, '#' protegido
    """
    n_attr  = len(attrs)
    n_meth  = len(methods)
    n_rows  = n_attr + n_meth + (1 if attrs and methods else 0)  # +1 para separador
    total_h = ROW_H * (n_rows + 1.6)   # 1.6 filas para la cabecera
    bottom  = y - total_h

    # Sombra
    ax.add_patch(FancyBboxPatch((x + 0.10, bottom - 0.10), w, total_h,
        boxstyle='round,pad=0.04', fc='#90A4AE', ec='none', alpha=0.20, zorder=1))
    # Cuerpo
    ax.add_patch(FancyBboxPatch((x, bottom), w, total_h,
        boxstyle='round,pad=0.04', fc=bg, ec=hc, lw=2.2, zorder=2))

    # Cabecera
    hdr_h = ROW_H * 1.6
    ax.add_patch(mpatches.Rectangle((x, y - hdr_h), w, hdr_h,
        fc=hc, ec='none', zorder=3))
    if stereotype:
        ax.text(x + w/2, y - hdr_h/2 + ROW_H * 0.35, f'«{stereotype}»',
            ha='center', va='center', fontsize=6.5, color='white',
            style='italic', zorder=4)
        ax.text(x + w/2, y - hdr_h/2 - ROW_H * 0.15, name,
            ha='center', va='center', fontsize=9, fontweight='bold', color='white', zorder=4)
    else:
        ax.text(x + w/2, y - hdr_h/2, name,
            ha='center', va='center', fontsize=9.5, fontweight='bold', color='white', zorder=4)

    # Atributos
    cur_y = y - hdr_h
    for i, (vis, aname, atype) in enumerate(attrs):
        row_y = cur_y - (i + 1) * ROW_H
        ax.plot([x + 0.06, x + w - 0.06], [row_y + ROW_H, row_y + ROW_H],
            color=hc, lw=0.35, alpha=0.30, zorder=3)
        vis_color = DARK_BLUE if vis == '+' else STEEL
        ax.text(x + 0.12, row_y + ROW_H * 0.5, vis,
            ha='left', va='center', fontsize=8, fontweight='bold', color=vis_color, zorder=4)
        ax.text(x + 0.32, row_y + ROW_H * 0.5, aname,
            ha='left', va='center', fontsize=6.8, color=DARK_TEXT, zorder=4)
        ax.text(x + w - 0.08, row_y + ROW_H * 0.5, f': {atype}',
            ha='right', va='center', fontsize=6.3, color='#757575',
            style='italic', zorder=4)

    # Separador atributos / métodos
    sep_y = cur_y - (n_attr + 1) * ROW_H
    if attrs and methods:
        ax.plot([x + 0.06, x + w - 0.06], [sep_y + ROW_H, sep_y + ROW_H],
            color=hc, lw=1.5, alpha=0.60, zorder=3)

    # Métodos
    meth_offset = n_attr + (1 if attrs and methods else 0)
    for j, (vis, mname, mtype) in enumerate(methods):
        row_y = cur_y - (meth_offset + j + 1) * ROW_H
        ax.plot([x + 0.06, x + w - 0.06], [row_y + ROW_H, row_y + ROW_H],
            color=hc, lw=0.35, alpha=0.30, zorder=3)
        ax.text(x + 0.12, row_y + ROW_H * 0.5, vis,
            ha='left', va='center', fontsize=8, fontweight='bold', color=DARK_BLUE, zorder=4)
        ax.text(x + 0.32, row_y + ROW_H * 0.5, f'{mname}()',
            ha='left', va='center', fontsize=6.8, color=DARK_TEXT,
            style='italic', zorder=4)
        if mtype:
            ax.text(x + w - 0.08, row_y + ROW_H * 0.5, f': {mtype}',
                ha='right', va='center', fontsize=6.3, color='#757575',
                style='italic', zorder=4)

    cx = x + w / 2
    return {'cx': cx, 'top': y, 'bottom': bottom,
            'left': x, 'right': x + w, 'mid_y': (y + bottom) / 2}

# ── Figura ────────────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(24, 16))
ax.set_xlim(0, 24)
ax.set_ylim(0, 16)
ax.axis('off')
ax.set_facecolor('#F5F7FA')
fig.patch.set_facecolor('#F5F7FA')

# ── Título ────────────────────────────────────────────────────────────────────
ax.add_patch(mpatches.Rectangle((0, 15.35), 24, 0.65, fc=NAVY, ec='none', zorder=0))
ax.text(12.0, 15.67, 'DIAGRAMA DE CLASES — MODELO DE DOMINIO SENDA',
    ha='center', va='center', fontsize=14, fontweight='bold', color='white', zorder=1)
ax.text(12.0, 15.40, 'TypeScript / Node.js · Supabase Self-Hosted · UNADECA 2026',
    ha='center', va='center', fontsize=8.5, color='#B0BEC5', zorder=1)

# ── Clases ────────────────────────────────────────────────────────────────────
C = {}

# Fila 1: Enums (parte superior)
C['UserRole'] = class_box(ax, 0.4, 15.1, 'UserRole', 'enumeration',
    attrs=[
        ('+', 'SUPER_ADMIN',  'string'),
        ('+', 'ADMIN',        'string'),
        ('+', 'DEPT_HEAD',    'string'),
        ('+', 'STUDENT',      'string'),
        ('+', 'ACCOUNTING',   'string'),
    ], methods=[],
    hc=NAVY, bg=WHITE_BLUE, w=4.2)

C['WorkLogStatus'] = class_box(ax, 5.0, 15.1, 'WorkLogStatus', 'enumeration',
    attrs=[
        ('+', 'PENDING',   'string'),
        ('+', 'APPROVED',  'string'),
        ('+', 'REJECTED',  'string'),
        ('+', 'PROCESSED', 'string'),
    ], methods=[],
    hc=DEEP_BLUE, bg=PALE_BLUE, w=4.2)

# Fila 2: Clases principales
C['Profile'] = class_box(ax, 0.4, 11.6, 'Profile', None,
    attrs=[
        ('-', 'id',                   'UUID'),
        ('-', 'email',                'string'),
        ('-', 'full_name',            'string'),
        ('-', 'role',                 'UserRole'),
        ('-', 'department_id',        'UUID | null'),
        ('-', 'is_active',            'boolean'),
        ('-', 'must_change_password', 'boolean'),
        ('-', 'hourly_rate_override', 'number | null'),
    ],
    methods=[
        ('+', 'getWorkLogs',     'WorkLog[]'),
        ('+', 'changePassword',  'void'),
    ],
    hc=DARK_BLUE, bg=LIGHT_BLUE, w=4.6)

C['Rate'] = class_box(ax, 5.4, 11.6, 'Rate', None,
    attrs=[
        ('-', 'id',             'UUID'),
        ('-', 'hourly_rate',    'number'),
        ('-', 'effective_date', 'Date'),
        ('-', 'created_by',     'UUID'),
    ],
    methods=[
        ('+', 'getEffective', 'Rate'),
    ],
    hc=SLATE_BLUE, bg=WHITE_BLUE, w=4.4)

C['Department'] = class_box(ax, 10.2, 11.6, 'Department', None,
    attrs=[
        ('-', 'id',              'UUID'),
        ('-', 'name',            'string'),
        ('-', 'cost_center',     'string'),
        ('-', 'kiosk_active',    'boolean'),
        ('-', 'kiosk_shift_start','string'),
        ('-', 'kiosk_shift_end', 'string'),
    ],
    methods=[
        ('+', 'toggleKiosk', 'void'),
        ('+', 'getMembers',  'Profile[]'),
    ],
    hc=ROYAL_BLUE, bg=LIGHT_BLUE, w=4.6)

C['WorkLog'] = class_box(ax, 15.2, 11.6, 'WorkLog', None,
    attrs=[
        ('-', 'id',           'UUID'),
        ('-', 'student_id',   'UUID'),
        ('-', 'department_id','UUID'),
        ('-', 'start_time',   'Date'),
        ('-', 'end_time',     'Date | null'),
        ('-', 'hours_worked', 'number'),
        ('-', 'status',       'WorkLogStatus'),
        ('-', 'description',  'string'),
        ('-', 'approved_by',  'UUID | null'),
    ],
    methods=[
        ('+', 'approve',   'void'),
        ('+', 'reject',    'void'),
        ('+', 'process',   'void'),
    ],
    hc=INDIGO, bg=WHITE_BLUE, w=4.8)

# Fila 3: Clases de soporte
C['KioskSession'] = class_box(ax, 0.4, 5.2, 'KioskSession', None,
    attrs=[
        ('-', 'id',            'UUID'),
        ('-', 'student_id',    'UUID'),
        ('-', 'department_id', 'UUID'),
        ('-', 'clock_in',      'Date'),
        ('-', 'clock_out',     'Date | null'),
        ('-', 'work_log_id',   'UUID | null'),
    ],
    methods=[
        ('+', 'clockOut',      'WorkLog'),
    ],
    hc='#0277BD', bg=SKY_BLUE, w=4.6)

C['AccountingConfig'] = class_box(ax, 5.4, 5.2, 'AccountingConfig', None,
    attrs=[
        ('-', 'id',             'UUID'),
        ('-', 'department_id',  'UUID'),
        ('-', 'cost_center',    'string'),
        ('-', 'account_debit',  'string'),
        ('-', 'account_credit', 'string'),
        ('-', 'tithe_account',  'string'),
        ('-', 'closing_day',    'number'),
    ],
    methods=[
        ('+', 'generateReport', 'Report'),
        ('+', 'exportTxt',      'string'),
    ],
    hc='#01579B', bg=SKY_BLUE, w=4.8)

# ── Relaciones (flechas de asociación / dependencia) ──────────────────────────
def rel(ax, c1, c2, verb, from_s, to_s, c, rad=0.0, card=''):
    b1, b2 = C[c1], C[c2]
    pts = {
        'top':    lambda b: (b['cx'], b['top']),
        'bottom': lambda b: (b['cx'], b['bottom']),
        'left':   lambda b: (b['left'], b['mid_y']),
        'right':  lambda b: (b['right'], b['mid_y']),
    }
    sx, sy = pts[from_s](b1)
    ex, ey = pts[to_s](b2)
    ax.annotate('', xy=(ex, ey), xytext=(sx, sy),
        arrowprops=dict(arrowstyle='->', color=c, lw=1.8,
            connectionstyle=f'arc3,rad={rad}'))
    mx, my = (sx + ex) / 2, (sy + ey) / 2
    if verb:
        ax.text(mx + 0.12, my + 0.14, verb,
            fontsize=6.8, color=c, style='italic', ha='left', va='bottom', zorder=5)
    if card:
        ax.text(ex + 0.10, ey + 0.10, card,
            fontsize=7.5, fontweight='bold', color=c, ha='left', va='bottom', zorder=5)

rel(ax, 'Profile',    'UserRole',      'usa',          'top',    'bottom', NAVY,    rad=0.0)
rel(ax, 'WorkLog',    'WorkLogStatus', 'usa',          'top',    'bottom', DEEP_BLUE, rad=0.2)
rel(ax, 'Profile',    'Department',    'pertenece a',  'right',  'left',   STEEL,   rad=0.0, card='0..N → 1')
rel(ax, 'WorkLog',    'Profile',       'realizado por','left',   'right',  DARK_BLUE, rad=-0.15, card='N → 1')
rel(ax, 'WorkLog',    'Department',    'en depto.',    'top',    'bottom', ROYAL_BLUE, rad=0.2)
rel(ax, 'KioskSession','Profile',      'del estudiante','right', 'left',   SLATE_BLUE, rad=0.12, card='N → 1')
rel(ax, 'KioskSession','Department',   'en depto.',    'bottom', 'bottom', STEEL,   rad=0.35)
rel(ax, 'AccountingConfig','Department','configura',   'right',  'bottom', '#0277BD', rad=-0.2)

# ── Leyenda ───────────────────────────────────────────────────────────────────
leg_x, leg_y = 20.8, 10.5
ax.add_patch(FancyBboxPatch((leg_x, leg_y), 2.9, 2.4,
    boxstyle='round,pad=0.10', fc='white', ec=NAVY, lw=1.2, zorder=4))
ax.text(leg_x + 1.45, leg_y + 2.18, 'Visibilidad UML',
    ha='center', fontsize=8.5, fontweight='bold', color=NAVY, va='center')
leg_items = [
    ('+', 'público',    DARK_BLUE),
    ('-', 'privado',    STEEL),
    ('#', 'protegido',  INDIGO),
    ('→', 'asociación', SLATE_BLUE),
]
for j, (sym, desc, col) in enumerate(leg_items):
    ax.text(leg_x + 0.20, leg_y + 1.82 - j * 0.45,
        f'{sym}  {desc}',
        ha='left', va='center', fontsize=8, color=col, fontweight='bold')

# ── Pie de página ─────────────────────────────────────────────────────────────
fig.text(0.5, 0.005,
    'Sistema SENDA — UNADECA 2026  |  Documento de Trabajo de Graduación',
    ha='center', fontsize=7.5, color=FOOTER, style='italic')

ax.set_title('Figura 7. Diagrama de Clases — Modelo de Dominio SENDA (UNADECA, 2026)',
    fontsize=13, fontweight='bold', pad=8)

save(fig, '7_clases.png')
```
